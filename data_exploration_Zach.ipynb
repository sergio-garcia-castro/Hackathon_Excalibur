```python
def preprocess_df(df):
    dataframe = df.copy()
    dataframe[['patient_id', 'date']] = dataframe['recording_id'].str.split('/', expand=True)

    #mettre patient et date en premier
    cols = dataframe.columns.tolist()
    cols = ['patient_id', 'date'] + [col for col in cols if col not in ['patient_id', 'date']]
    dataframe = dataframe[cols]

    dataframe = dataframe.drop(columns=['recording_id'])

    #enlever 'patient_' et tous les 0 avant de la colonne patient_id sauf si c'est le seul caractère
    dataframe['patient_id'] = dataframe['patient_id'].str.replace('patient_', '', regex=False).str.lstrip('0')
    dataframe['patient_id'] = dataframe['patient_id'].replace('', '0')

    return dataframe

def str_dict_to_dict(s):
    import ast
    try:
        return ast.literal_eval(s)
    
    except (ValueError, SyntaxError):
        return {}  # Return an empty dictionary if conversion fails
```

```python
dataset = []

for patient in test_df['patient_id'].unique():
    print (f"Processing patient {patient}")
    patient_data = test_df[test_df['patient_id'] == patient]
    for date in patient_data['date'].unique():
        date_data = patient_data[patient_data['date'] == date]
        
        for recording_id in date_data['recording_id'].unique():
            rec_data = date_data[date_data['recording_id'] == recording_id]
            element = {}
            element['patient_id'] = patient
            element['date'] = date
            #trier rec_data par start_time croissant
            rec_data = rec_data.sort_values(by='start_time')
            embeddings = []
            start_times = []
            end_times = []
            labels = []
            for i, row in rec_data.iterrows():
                #récuperer toutes les colonnes sauf patient_id, date, recording_id, start_time, end_time et les valeurs non numériques sous forme de liste
                embedding = row.drop(labels=['patient_id','patient_short_id','label', 'date', 'recording_id', 'start_time', 'end_time']).tolist()
                #print(f'Patient {patient}, embedding length: {len(embedding)}')
                embeddings.append(embedding)
                start_times.append(row['start_time'])
                end_times.append(row['end_time'])
                labels.append(row['label'])
            
            element['embeddings'] = embeddings
            element['start_times'] = start_times
            element['end_times'] = end_times
            element['label'] = labels[0]

            dataset.append(element)
```

```python
import numpy as np
from datetime import datetime

positive_pairs = []
negative_pairs = []

i = 0
for data1 in dataset:

    if i % 100 == 0:
        print(f"Processing record {i} / {len(dataset)}")
    
    i += 1
    patient_id = data1['patient_id']
    date = data1['date']
    embeddings = data1['embeddings']

    #récupérer toutes les autres dates pour le même patient
    other_dates = [d['date'] for d in dataset if d['patient_id'] == patient_id and d['date'] != date]

    #récupérer si elles existent les dates les plus proches avant et après
    date_obj = datetime.strptime(date, '%Y-%m-%d')
    before_dates = [d for d in other_dates if datetime.strptime(d, '%Y-%m-%d') < date_obj]
    after_dates = [d for d in other_dates if datetime.strptime(d, '%Y-%m-%d') > date_obj]
    closest_before = max(before_dates, key=lambda d: datetime.strptime(d, '%Y-%m-%d')) if before_dates else None
    closest_after = min(after_dates, key=lambda d: datetime.strptime(d, '%Y-%m-%d')) if after_dates else None

    for data2 in dataset :
        if data2['patient_id'] == patient_id and data2['date'] in [closest_before, date_obj, closest_after]:
            positive_pairs.append([embeddings, data2['embeddings']])
            
        else :
            negative_pairs.append([embeddings, data2['embeddings']])

pairs = []

for pos in positive_pairs:
    pairs.append([pos[0], pos[1], 1])  # 1 for positive pair

for neg in negative_pairs:
    pairs.append([neg[0], neg[1], 0])  # 0 for negative pair
```

```python
import base64
import gzip
from contextlib import contextmanager
from dataclasses import dataclass
from typing import Dict, Iterable, Optional, Tuple

import numpy as np
import torch
import torch.nn.functional as F
from torch import Tensor, nn
```

```python
class PositionEncoding(nn.Module):
    def __init__(self, d_model=784, max_len=100):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float32).unsqueeze(1)
        embedding_index = torch.arange(0, d_model, 2, dtype=torch.float32)
        div_term = 1 / torch.tensor(10000.0) ** (embedding_index / d_model)
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        self.register_buffer('pe', pe)  # (max_len, d_model)

    def forward(self, word_embeddings):
        # accepts (batch, seq_len, d_model) or (seq_len, d_model)
        if word_embeddings.dim() == 3:
            seq_len = word_embeddings.size(1)
            pe = self.pe[:seq_len].unsqueeze(0).to(word_embeddings.device).to(word_embeddings.dtype)  # (1, seq_len, d_model)
            return word_embeddings + pe
        else:
            seq_len = word_embeddings.size(0)
            pe = self.pe[:seq_len].to(word_embeddings.device).to(word_embeddings.dtype)  # (seq_len, d_model)
            return word_embeddings + pe


    
class SelfMultiheadAttention(nn.Module):
    def __init__(self, d_emb=784, nhead=8,d_out=None):
        super().__init__()
        assert d_emb % nhead == 0, "d_emb must be divisible by nhead"
        
        self.d_model = d_emb // nhead
        self.nhead = nhead
        self.PositionalEncoding = PositionEncoding(d_model=d_emb,max_len=100)
        self.Q = nn.Linear(d_emb, d_emb) # d_emb x d_emb
        self.K = nn.Linear(d_emb, d_emb) # d_emb x d_emb
        self.V = nn.Linear(d_emb, d_emb) # d_emb x d_emb

        if d_out is None:
            d_out = d_emb

        self.out = nn.Linear(d_emb, d_out) # d_emb x d_out
    
    def forward(self, x, mask=None):

        batch_size, seq_len, d_emb = x.size()

        Q = self.Q(x).view(batch_size, seq_len, self.nhead, self.d_model).transpose(1, 2)  # batch_size x nhead x seq_len x d_model
        K = self.K(x).view(batch_size, seq_len, self.nhead, self.d_model).transpose(1, 2)  # batch_size x nhead x seq_len x d_model
        V = self.V(x).view(batch_size, seq_len, self.nhead, self.d_model).transpose(1, 2)  # batch_size x nhead x seq_len x d_model

        scores = torch.matmul(Q, K.transpose(-2, -1)) / np.sqrt(self.d_model) 
        if mask is not None:
            scores = scores.masked_fill(mask == 0, float('-inf'))
        attn_weights = F.softmax(scores, dim=-1)

        attn_output = torch.matmul(attn_weights, V)
        attn_output = attn_output.transpose(1, 2).contiguous().view(batch_size, seq_len, d_emb)

        logits = self.out(attn_output)

        return logits
    

class TransformerEncoderLayer(nn.Module):
    def __init__(self, d_model=784, nhead=8, dropout=0.1):
        super().__init__()
        dim_feedforward = d_model * 4
        self.self_attn = SelfMultiheadAttention(d_model, nhead, d_out=d_model)
        self.linear1 = nn.Linear(d_model, dim_feedforward)
        self.dropout = nn.Dropout(dropout)
        self.linear2 = nn.Linear(dim_feedforward, d_model)

        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.dropout1 = nn.Dropout(dropout)
        self.dropout2 = nn.Dropout(dropout)

        self.activation = F.relu

    def forward(self, src,mask=None):
        src2 = self.self_attn(src,mask=mask)
        src = src + self.dropout1(src2)
        src = self.norm1(src)
        src2 = self.linear2(self.dropout(self.activation(self.linear1(src))))
        src = src + self.dropout2(src2)
        src = self.norm2(src)
        return src

class SelfMultiheadAttentionPoolingProjector(nn.Module):
    def __init__(self, d_emb=784, nhead=8):
        super().__init__()
        assert d_emb % nhead == 0, "d_emb must be divisible by nhead"
        
        self.d_model = d_emb // nhead
        self.nhead = nhead
        self.Attention = SelfMultiheadAttention(d_emb=d_emb, nhead=nhead, d_out=1)

    def forward(self, x, mask=None, temp=1.0):

        logits = self.Attention(x, mask=mask)

        weights = F.softmax(logits.squeeze()/temp)

        # weighted sum of the input embeddings
        logits = torch.sum(x * weights.unsqueeze(-1), dim=1)


        return logits,weights
    
class Embedder(nn.Module):
    def __init__(self, d_emb=784, nhead=8, num_transformer_layers=2,num_mlp_layers=2, d_projector=None):
        super().__init__()
        if d_projector is None:
            d_projector = d_emb
            
        self.d_projector = d_projector
        self.d_model = d_projector // nhead

        self.nhead = nhead
        self.num_transformer_layers = num_transformer_layers
        self.num_mlp_layers = num_mlp_layers

        self.InitialProjector = nn.Linear(d_emb, d_projector)
        self.PositionalEncoding = PositionEncoding(d_model=d_projector,max_len=1000)
        self.TransformerLayers = nn.ModuleList([
            TransformerEncoderLayer(d_model=d_projector, nhead=nhead) for _ in range(num_transformer_layers)
        ])

        self.AttentionPooling = SelfMultiheadAttentionPoolingProjector(d_emb=d_projector, nhead=nhead)

        self.MLP_layers = nn.ModuleList(
            nn.Sequential(
                nn.Linear(d_projector, d_projector),
                nn.ReLU(),
                nn.Linear(d_projector, d_projector)
            ) for _ in range(num_mlp_layers)
        )

    def forward(self, x, mask=None):

        x = self.InitialProjector(x)
        
        x = self.PositionalEncoding(x)

        for layer in self.TransformerLayers:
            x = layer(x, mask=mask)

        x, weights = self.AttentionPooling(x, mask=mask)

        for mlp in self.MLP_layers:
            x = mlp(x)

        return x
    

    def __repr__(self):
        #take the number of parameters in the model
        n_params = sum(p.numel() for p in self.parameters() if p.requires_grad)
        return f"Embedder(d_emb={self.d_model * self.nhead}, nhead={self.nhead}, num_transformer_layers={len(self.TransformerLayers)}, num_mlp_layers={len(self.MLP_layers)}, n_params={n_params})"

        
```

```python
from torch.utils.data import DataLoader, Dataset
```

```python
class EmbeddingDataset(Dataset):
    """
    Dataset contenant des paires (z1, z2) et un label.
    Chaque z est une liste de vecteurs (seq_len, emb_dim)
    """
    def __init__(self, data):
        """
        data : list of tuples (z1, z2, label)
               z1 and z2 : list of embeddings [seq_len, emb_dim]
               label : scalar
        """
        self.data = data

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        return self.data[idx]
```

```python
def embedding_collate_fn(batch):
    """
    Transforme une liste de samples en batch avec padding et masques.
    Renvoie le mask prêt pour SelfMultiheadAttention.
    """
    z_1_list = [torch.tensor(x[0], dtype=torch.float) for x in batch]
    z_2_list = [torch.tensor(x[1], dtype=torch.float) for x in batch]
    labels = torch.tensor([x[2] for x in batch], dtype=torch.float)

    # longueur max par batch
    max_len_z1 = max([z.shape[0] for z in z_1_list])
    max_len_z2 = max([z.shape[0] for z in z_2_list])
    emb_dim = z_1_list[0].shape[1]  # dimension embedding

    # batch padded
    z1_padded = torch.zeros((len(batch), max_len_z1, emb_dim))
    z2_padded = torch.zeros((len(batch), max_len_z2, emb_dim))
    
    # attention mask pour le batch
    attention_mask_z1 = torch.zeros((len(batch), 1, 1, max_len_z1), dtype=torch.bool)
    attention_mask_z2 = torch.zeros((len(batch), 1, 1, max_len_z2), dtype=torch.bool)

    for idx, z in enumerate(z_1_list):
        seq_len = z.shape[0]
        z1_padded[idx, :seq_len, :] = z
        attention_mask_z1[idx, :, :, :seq_len] = 1  # shape: (B,1,1,L)

    for idx, z in enumerate(z_2_list):
        seq_len = z.shape[0]
        z2_padded[idx, :seq_len, :] = z
        attention_mask_z2[idx, :, :, :seq_len] = 1

    return {
        "z_1": z1_padded,
        "attention_mask_z1": attention_mask_z1,
        "z_2": z2_padded,
        "attention_mask_z2": attention_mask_z2,
        "labels": labels
    }



def get_embedding_dataloader(dataset_list, batch_size=8, shuffle=True):
    """
    Crée un DataLoader PyTorch pour les embeddings
    """
    dataset = EmbeddingDataset(dataset_list)
    dataloader = DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=shuffle,
        collate_fn=embedding_collate_fn
    )
    return dataloader
```

```python
import torch.nn.functional as F

def contrastive_loss(emb1, emb2, labels, temperature=0.5):
    """
    emb1, emb2: (batch_size, d_emb)
    labels: 1 si similaire, 0 sinon
    """
    # Normalisation
    emb1 = F.normalize(emb1, dim=1)
    emb2 = F.normalize(emb2, dim=1)

    # Similarité cosine
    sim = torch.sum(emb1 * emb2, dim=1) / temperature  # (batch_size,)

    # Labels 1 -> positive, 0 -> negative
    loss = F.binary_cross_entropy_with_logits(sim, labels)
    return loss
```

```python
# Device
device = torch.device("cpu")

# Embedder
embedder = Embedder(d_emb=784,
                     nhead=8,
                     num_transformer_layers=0,
                     num_mlp_layers=0,
                     d_projector=64).to(device)

# Optimizer
optimizer = torch.optim.Adam(embedder.parameters(), lr=1e-4)
```

```python
# Loop d'entraînement
for epoch in range(1):  # petit test
    embedder.train()
    total_loss = 0
    i = 0
    for batch in dataloader:
        if i % 100 == 0:
            print(f"Processing batch {i+1}/{len(dataloader)}")
        i += 1
        Z1 = batch["z_1"].to(device)
        Z2 = batch["z_2"].to(device)
        mask1 = batch["attention_mask_z1"].to(device)
        mask2 = batch["attention_mask_z2"].to(device)
        labels = batch["labels"].to(device)

        optimizer.zero_grad()
        emb1 = embedder(Z1, mask=mask1)
        emb2 = embedder(Z2, mask=mask2)

        loss = contrastive_loss(emb1, emb2, labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
    
    print(f"Epoch {epoch+1}, Loss: {total_loss/len(dataloader):.4f}")
```
